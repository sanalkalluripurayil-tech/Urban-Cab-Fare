```python
# Replacing null values
df["Driver_Name"] = df["Driver_Name"].fillna("Not Mentioned")
df.isnull().sum()
```

```python
print("Number of Duplicate Values: ", df.duplicated().sum())
```

```python
# Changing datatype
df["Ride_Date"] = pd.to_datetime(df["Ride_Date"], dayfirst=True)
df.dtypes
```

```python
# adding new colomn Distance_Category
df['Distance_Category'] = np.select(
    [
        df['Distance_km'] < 10,
        df['Distance_km'].between(10, 20),
        df['Distance_km'] > 20
    ],
    ['Short', 'Medium', 'Long'],
    default='Unknown'
)

# adding new column Year_Month
df['Year_Month'] = df['Ride_Date'].dt.to_period('M')

df
```

```python
#duration
starting_date = df["Ride_Date"].min()
print("Starting Date:", starting_date)

ending_date = df["Ride_Date"].max()
print("Ending Date:", ending_date)

duration = (ending_date - starting_date).days
print("Duration:", duration)
```

```python
# grouping revenue by month
monthly_fotal_fare = df.groupby('Year_Month')['Final_Fare'].sum()

# convert Series to DataFrame
monthly_df = monthly_fotal_fare.reset_index()

# converting year_month to str from period
monthly_df['Year_Month'] = monthly_df['Year_Month'].astype(str)

# Create line chart
fig = px.area(monthly_df, x="Year_Month", y="Final_Fare",
              title="Monthly Revenue Trend",
              markers=False)

fig.update_traces(line_color="lightgreen")

fig.update_layout(
    width=1000,
    height=500,
    xaxis_title="Month",
    yaxis_title="Revenue",
    plot_bgcolor="white")

fig.show()

top_3_months = monthly_fotal_fare.nlargest(3)

print(top_3_months)
```

```python
monthly_total_rides = df.groupby('Year_Month')[['Ride_Requests', 'No_of_active_drivers']].sum()

fig = px.bar(
    monthly_total_rides,
    x=monthly_total_rides.index.astype(str),
    y=["Ride_Requests", "No_of_active_drivers"],
    barmode="group",
    title="Ride Requests vs Active Drivers by Month",
    color_discrete_sequence=["blue", "lightgreen"]
)

fig.update_traces(opacity=1)

fig.update_layout(
    width=1100,
    height=500,
    xaxis_title="Month",
    yaxis_title="Count",
    plot_bgcolor="white",
)

fig.show()

top_3_months = monthly_total_rides.nlargest(3, "Ride_Requests")
print(top_3_months)
```

```python
# Calculate total revenue by city
Total_fare_by_city = (
    df.groupby('City')['Final_Fare']
      .sum()
      .sort_values(ascending=True)
)

# Create horizontal bar chart
fig = px.bar(
    x=Total_fare_by_city.values,
    y=Total_fare_by_city.index.astype(str),
    orientation='h',
    title="Total Revenue by City",
    labels={
        'x': 'Total Revenue',
        'y': 'City'
    }
)

fig.update_traces(
    marker_color="lightgreen"
)

fig.update_layout(
    width=700,
    height=500,
    plot_bgcolor="white"
)

fig.show()

# Find the city with the highest revenue
highest_city = Total_fare_by_city.idxmax()
highest_revenue = Total_fare_by_city.max()

# Find the city with the lowest revenue
lowest_city = Total_fare_by_city.idxmin()
lowest_revenue = Total_fare_by_city.min()

print("City With Highest Revenue:")
print(highest_city, highest_revenue)

print("\nCity With Lowest Revenue:")
print(lowest_city, lowest_revenue)
```

```python
# Calculate ride requests and active drivers by city
active_drives_by_city = (
    df.groupby("City")[["Ride_Requests", "No_of_active_drivers"]]
      .sum()
      .sort_values(by="Ride_Requests", ascending=True)
)

# Create horizontal grouped bar chart
fig = px.bar(
    active_drives_by_city,
    x=["No_of_active_drivers", "Ride_Requests"],
    y=active_drives_by_city.index.astype(str),
    barmode="group",
    orientation="h",
    title="Ride Requests vs Active Drivers by City",
    color_discrete_sequence=["blue", "lightgreen"]
)

fig.update_layout(
    width=1000,
    height=650,
    xaxis_title="Count",
    yaxis_title="City",
    plot_bgcolor="white"
)

fig.show()

# Highest ride requests
highest_city = active_drives_by_city["Ride_Requests"].idxmax()
highest_request = active_drives_by_city["Ride_Requests"].max()

# Lowest ride requests
lowest_city = active_drives_by_city["Ride_Requests"].idxmin()
lowest_request = active_drives_by_city["Ride_Requests"].min()

print("City With Highest Ride Requests:")
print(highest_city, highest_request)

print("\nCity With Lowest Ride Requests:")
print(lowest_city, lowest_request)
```

```python
# Create treemap
average_surge_by_city = df.groupby("City")["Surge_Multiplier"].mean()

plt.figure(figsize=(10, 6))

squarify.plot(
    sizes=average_surge_by_city.values,
    label=[
        f"{city}\n{ratio:.2f}"
        for city, ratio in average_surge_by_city.items()
    ],
    alpha=0.7
)

plt.title("Average Surge-Multiplier by City")
plt.axis("off")
plt.show()
```

```python
heatmap_data = df.pivot_table(
    index='Day_of_Week',
    columns='Hour_of_Day',
    values='Ride_Requests',
    aggfunc='mean'
)

day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

heatmap_data = heatmap_data.reindex(day_order)

plt.figure(figsize=(14, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt='.0f',
    cmap='YlOrRd'
)

plt.title("Average Ride Requests by Day and Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Day of Week")

plt.show()

top_5 = (
    df.groupby(["Day_of_Week", "Hour_of_Day"])["Ride_Requests"]
    .mean()
    .sort_values(ascending=False)
    .head(5)
)

print(top_5)
```

```python
# Filter only Jam traffic
jam_traffic = df[df['Traffic_Level'] == 'Jam']

# Count Jam traffic records by hour
traffic_by_hour = jam_traffic['Hour_of_Day'].value_counts().sort_index()

# Create area chart
fig = px.area(
    x=traffic_by_hour.index,
    y=traffic_by_hour.values,
    title="Peak Jam Traffic Hours",
    labels={
        "x": "Hour of Day",
        "y": "Number of Jam Traffic Records"
    }
)

fig.update_layout(
    width=1100,
    height=500,
)

fig.update_traces(
    line_color="red"
)
        
# Show chart
fig.show()

# Find peak hour
peak_hour = traffic_by_hour.idxmax()
peak_count = traffic_by_hour.max()

print("Peak Jam Traffic Hour:", peak_hour)
print("Jam Traffic Records:", peak_count)
```

```python
from matplotlib.colors import ListedColormap

traffic_day_city = (
    df.groupby(["City", "Day_of_Week"])["Traffic_Level"]
      .apply(lambda x: x.mode().iloc[0])
      .reset_index()
)

heatmap_data = traffic_day_city.pivot(
    index="City",
    columns="Day_of_Week",
    values="Traffic_Level"
)

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

heatmap_data = heatmap_data.reindex(columns=day_order)

# Convert traffic levels to category codes only for coloring
traffic_order = ["Low", "Medium", "High", "Jam"]

color_data = heatmap_data.apply(
    lambda col: col.map(
        {level: i for i, level in enumerate(traffic_order)}
    )
)

# Green → Yellow → Orange → Red
cmap = ListedColormap([
    "lightgreen",
    "yellow",
    "orange",
    "red"
])

plt.figure(figsize=(14, 6))

sns.heatmap(
    color_data,
    annot=heatmap_data,
    fmt="",
    cmap=cmap,
    cbar=False,
    linewidths=0.5,
    linecolor="white"
)

plt.title("Most Common Traffic Level by City and Day")
plt.xlabel("Day of Week")
plt.ylabel("City")

plt.xticks(rotation=45)
plt.yticks(rotation=0)

plt.show()
```

```python
import plotly.graph_objects as go

# Convert traffic levels into numeric values
traffic_mapping = {
    "Low": 1,
    "Medium": 2,
    "High": 3,
    "Jam": 4
}

df["Traffic_Level_Num"] = df["Traffic_Level"].map(traffic_mapping)

# Group data by hour
hourly_data = df.groupby("Hour_of_Day").agg(
    Traffic_Level=("Traffic_Level_Num", "mean"),
    Ride_Requests=("Ride_Requests", "sum")
).sort_index()

# Create chart with secondary Y-axis
fig = go.Figure()

# Line 1: Traffic level
fig.add_trace(
    go.Scatter(
        x=hourly_data.index,
        y=hourly_data["Traffic_Level"],
        mode="lines+markers",
        name="Average Traffic Level",
        line=dict(color="red")
    )
)

# Line 2: Ride requests
fig.add_trace(
    go.Scatter(
        x=hourly_data.index,
        y=hourly_data["Ride_Requests"],
        mode="lines+markers",
        name="Total Ride Requests",
        line=dict(color="blue"),
        yaxis="y2"
    )
)

fig.update_layout(
    title="Traffic Levels and Ride Requests by Hour",
    width=1100,
    height=500,
    plot_bgcolor="white",
    xaxis=dict(title="Hour of Day", dtick=1),
    yaxis=dict(
        title="Average Traffic Level",
        tickmode="array",
        tickvals=[1, 2, 3, 4],
        ticktext=["Low", "Medium", "High", "Jam"]
    ),
    yaxis2=dict(
        title="Total Ride Requests",
        overlaying="y",
        side="right"
    ),
    hovermode="x unified"
)

fig.show()
```
